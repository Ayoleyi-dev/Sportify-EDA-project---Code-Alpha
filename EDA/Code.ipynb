```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Load the dataset
    df = pd.read_csv("spotify_tracks.csv", encoding='ISO-8859-1') # Note: used 'encoding="ISO-8859-1"' because Spotify datasets often have special characters
except FileNotFoundError:
    df = pd.read_csv("spotify_tracks.csv", encoding='ISO-8859-1')

# To Show the first few rows and the column names
print("--- COLUMN NAMES ---")
print(df.columns.tolist())

print("\n--- FIRST 5 ROWS ---")
display(df.head())

# Check for missing values
print("\n--- MISSING VALUES ---")
print(df.isnull().sum())
```

```python
# Convert ms to minutes
df["duration_min"] = df["duration_ms"] / 60000

print("New column created: duration_min")
display(df[["name", "duration_min"]].head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
#To Get the top 10 genres
top_genres = df["genre"].value_counts().head(10).index

# Filter data to only include top 10 for the plot
sns.countplot(data=df[df["genre"].isin(top_genres)], y="genre", order=top_genres, palette="viridis")

plt.title("Top 10 Genres by Number of Tracks")
plt.xlabel("Number of Tracks")
plt.ylabel("Genre")
plt.show()
```

```python
#For histogram visualization:

plt.figure(figsize=(8, 5))
sns.histplot(df["popularity"], bins=30, kde=True, color="purple")
plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity Score (0-100)")
plt.show()
```

```python
plt.figure(figsize=(5, 6))
df["explicit"].value_counts().plot(kind="pie", autopct="%1.1f%%", colors=["#F9C80E", "#EA3546"], labels=["Clean", "Explicit"])
plt.title("Explicit vs. Clean Content")
plt.ylabel("") # Hides the 'explicit' label on the side
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x="duration_min", y="popularity", data=df, alpha=0.3, color="blue")
plt.title("Does Song Length Affect Popularity?")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Popularity")
plt.xlim(0, 10) # Limit to songs under 10 minutes to hide outliers
plt.show()
```
